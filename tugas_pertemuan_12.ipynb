1. Luas Persegi Panjang

```python
def luas_pp(panjang, lebar):
  rumus_luas_pp = panjang * lebar
  return rumus_luas_pp

p = float(input('panjang = '))
l = float(input('lebar = '))
hasil_luas_pp = luas_pp(p, l)

print(f'Luas persegi panjang = {p} * {l} = {hasil_luas_pp}')
```

3. Luas Lingkaran

```python
import math

def luas_lingkaran (radius):
  rumus_luas_lingkaran = math.pi * radius ** 2
  return rumus_luas_lingkaran

r = float(input('jari-jari = '))
hasil_luas_lingkaran = luas_lingkaran(r)
print(f'Luas lingkaran = {math.pi} * {r} ** 2 = {hasil_luas_lingkaran:.2f}')
```

4. Volume Bola

```python
import math

def volume_bola(radius):
  rumus_volume_bola = 4/3 * math.pi * radius ** 3
  return rumus_volume_bola

r = float(input('jari-jari = '))
hasil_volume_bola = volume_bola(r)

print(f'Volume bola = 4/3 * {math.pi} * {r} ** 3 = {hasil_volume_bola:.2f}')
```

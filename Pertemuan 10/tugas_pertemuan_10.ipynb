Dictionary

```python
# Membuat dictionary
keranjang = {
    'apel' : 5000,
    'pisang' : 3000,
    'jeruk' : 7000,
    'anggur' : 15000
}

# Menghitung total harga

total = 0
for key, value in keranjang.items():
  total += value
  print(f'harga {key}: Rp {value}')
print(f'Total belanja adalah Rp {total}')
```

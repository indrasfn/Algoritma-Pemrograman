Sistem Manajemen Perpustakaan

```python
import datetime
import random
import string

perpustakaan_template = {
    'judul' : 'judul buku',
    'penulis' : 'nama penulis',
    'tahun_terbit' : 1111,
    'tanggal_pinjam' : datetime.datetime(1111, 11, 11),
    'status' : 'tersedia/tidak'
}

data_perpustakaan = {}

while True:
  print('Selamat Datang di Perpustakaan Nusa Ilmu')
  print('='*100)

  perpustakaan = dict.fromkeys(perpustakaan_template.keys())

  # data buku
  perpustakaan['judul'] = input('Judul Buku: ')
  perpustakaan['penulis'] = input('Nama Penulis: ')
  perpustakaan['tahun_terbit'] = int(input('Tahun Terbit: (YYYY)'))

  # tangga pinjam
  TAHUN_PINJAM = int(input('Tahun Pinjam: (YYYY)'))
  BULAN_PINJAM = int(input('Bulan Pinjam: (1-12)'))
  TANGGAL_PINJAM = int(input('Tanggal Pinjam: (1-31)'))
  perpustakaan['tanggal_pinjam'] = datetime.datetime(TAHUN_PINJAM, BULAN_PINJAM, TANGGAL_PINJAM)

  # status buku
  tersedia = input('Apakah buku ini tersedia? (y/n)').lower()
  if tersedia == 'y':
    perpustakaan['status'] = 'Tersedia'
  else:
    perpustakaan['status'] = 'Tidak Tersedia'

  KEY = ''.join((random.choice(string.ascii_uppercase) for i in range(6) ))

  data_perpustakaan.update({KEY: perpustakaan})

  print(f'{'KEY':<10} {'Judul Buku':<15} {'Nama Penulis':<15} {'Tahun Terbit':<15} {'Tanggal Pinjam':<17} {'Status Buku':<17}')
  print('-'*100)

  for key in data_perpustakaan:
    KEY = perpustakaan
    judul = data_perpustakaan[key] ['judul']
    penulis = data_perpustakaan[key] ['penulis']
    tahun_terbit = data_perpustakaan[key] ['tahun_terbit']
    tanggal_pinjam = data_perpustakaan[key] ['tanggal_pinjam'].strftime('%x')
    status = data_perpustakaan[key] ['status']

    print(f'{key:<10} {judul:<17} {penulis:<16} {tahun_terbit:<15} {tanggal_pinjam:<16} {status:<17}')
  print('\n')
  selesai = input('Apakah ingin menambah data lagi? (y/n)')
  if selesai.lower() == 'n':
    break
print('\nTerima Kasih telah Mengunjungi Perpustakaan Nusa Ilmu')
```
